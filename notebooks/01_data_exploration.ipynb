# MELD Dataset Exploration
## Context-Aware Emotional Transition Detection

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('.').resolve().parent))

import pandas as pd
import plotly.express as px
from utils.data_loader import prepare_dataset, download_meld_dataset
from utils.transitions import emotion_transition_matrix
```

```python
download_meld_dataset()
train_df, val_df, test_df = prepare_dataset()
all_df = pd.concat([train_df, val_df, test_df])
print(f'Total transition samples: {len(all_df)}')
all_df.head()
```

```python
fig = px.histogram(all_df, x='next_emotion', title='Next Emotion Distribution', template='plotly_dark')
fig.show()
```

```python
matrix = emotion_transition_matrix(all_df['transition'].tolist())
fig2 = px.imshow(matrix, labels=dict(x='To', y='From', color='Count'), title='Transition Heatmap')
fig2.show()
```
